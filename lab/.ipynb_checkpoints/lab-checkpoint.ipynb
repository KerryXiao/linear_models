# Hedonic Pricing

We often try to predict the price of an asset from its observable characteristics. This is generally called **hedonic pricing**: How do the unit's characteristics determine its market price?

In the lab folder, there are three options: housing prices in pierce_county_house_sales.csv, car prices in cars_hw.csv, and airbnb rental prices in airbnb_hw.csv. If you know of another suitable dataset, please feel free to use that one.

1. Clean the data and perform some EDA and visualization to get to know the data set.
2. Transform your variables --- particularly categorical ones --- for use in your regression analysis.
3. Implement an ~80/~20 train-test split. Put the test data aside.
4. Build some simple linear models that include no transformations or interactions. Fit them, and determine their RMSE and $R^2$ on the both the training and test sets. Which of your models does the best?
5. Make partial correlation plots for each of the numeric variables in your model. Do you notice any significant non-linearities? 
6. Include transformations and interactions of your variables, and build a more complex model that reflects your ideas about how the features of the asset determine its value. Determine its RMSE and $R^2$ on the training and test sets. How does the more complex model your build compare to the simpler ones?
7. Summarize your results from 1 to 6. Have you learned anything about overfitting and underfitting, or model selection?
8. If you have time, use the sklearn.linear_model.Lasso to regularize your model and select the most predictive features. Which does it select? What are the RMSE and $R^2$? We'll cover the Lasso later in detail in class.

```python
1. The dataset has 976 rows and 12 columns. Since there was no missing values, the cleaning wasn't too difficult. The Unnamed:0 was dropped because it
was just an index column. I converted No_of_Owners from text (1st, 2nd, 3rd) into a numeric variable so it became 1, 2, and 3 instead. The mean price was
about 741019. The median price is 666500. The mean mileage is around 41571. The model years range from 2011 to 2022. A few EDA findings show that newer
cars tend to cost more, higher mileage tends to reduce price, SUVs are the most expensive body type on average, diesel cars are more expensive on average
than petrol and petrol+cng in the sample. Makes like MG Motors, Kia, and Jeep have the highest average prices in the data. This matches the hedonic 
pricing idea where price is related to observable characteristics like age, mileage, fuel type, body type, and make.

2. Since categorical variables cannot be used as words for regression, they need to be transformed. I used One-hot encoding for Make, Color, Body_Type, 
Fuel_Type, Transmission, Transmission_Type, and Owners_Num. For the more complex model, I also created Age which is equal to max(Make_Year) + 1 - Make_Year.
Log_Mileage is equal to log(Mileage_Run) was also created. The above help because car values change with mileage and age in a curved and not perfectly 
straight way.

3. I used 80% training, 20% testing, and random_state equal to 42. The point is to build the model on the training set and then evaluate it on unseen data
from the test set. 

4. I decided to built three simple models with no interactions and no nonlinear transformations. The models were Numeric only, Numeric + a few categorical,
and All main effects. The Train RMSE in order is 308974, 221074, and 149824. The Test RMSE in order is 284941, 205329, and 143274. The Train R^2 in order is 
0.316, 0650, and 0.839. The Test R^2 in order is 0.292, 0.632, and 0.821. From the data above, the best simple model is All main effects. The interpretation
is that using only numerical variables leaves out too much information, adding a few categorical variables helps, and including all major observable 
characteristics improves prediction.

5. CCPR plots were generated for the numeric variables in the best simple model. Mileage_Run suggests there are signs the relationship is not perfectly
linear. Price seems to fall with mileage, but not at a constant rate. Make_Year is mostly positive, but there is some structure suggesting that treating
age more directly may work better than just the raw year. Seating_Capacity's effect isn't that clean and behaves differently across body types. Owners_Num
effect is small and stepped since it only takes 1, 2, or 3.

6. I built a more complex model with transformations and interactions. log(Price) was the response. Age, Age^2, Log_Mileage, and Log_Mileage^2 were 
variables. Some interactions included Age x Fuel_Type, Log_Mileage x Transmission_Type, and Body_Type x Fuel_Type. For the Best Simple
Model, the Train RMSE was 149824, the Test RMSE was 143274, the Train R^2 was 0.839, and the Test R^2 was 0.821. For the Complex Model, the Train RMSE
was 136660, the Test RMSE was 123320, the Train R^2 was 0.866, and the Test R^2 was 0.867. The more complex model did better on both train and test data
with lower RMSE and higher R^2. This shows that the added transformations and interactions were useful and not just overfitting noise.

7. The main takeaway is that a very simple model underfits because it misses important structure. A richer model can improve test performance when the 
extra terms are meaningful. The goal is not to make the model as complicated as possible. The goal is to make it complicated enough to capture the real
pattern without chasing random noise. Underfitting happened in the numeric-only model. By adding relevant categorical variables a large improvement is seen.
Nonlinear terms and interactions improved things further. The complex model is not memorizing the training set because test performance improved too. This
is a good example of model selection because I started simple, fixed weaknesses, added structure that made sense, and compared train and test results.

8. I also fit a Lasso model using transformed numeric terms, polynomial terms, and one-hot encodede categorical variables. The model's results for Train
RMSE was 140556, Test RMSE was 132120, Train R^2 was 0.858, and Test R^2 was 0.848. Some of the strongest features Lasso kept Seating_Capacity^2; Seating_Capacity;
Log_Mileage x Seating_Capacity; Age x Log_Mileage; Log_Mileage^2; brand indicators like Make_MG Motors, Make_Hyundai, Make_Toyota, Make_Honda; body type
indicators like hatchback, suv, muv; Fuel_Type_petrol; and transmission and color indicators. Lasso essentially does feature selection by shrinking weaker
variables towards zero and keeping the strongest predictors. 
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.formula.api as smf

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, PolynomialFeatures, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression, LassoCV
from sklearn.metrics import mean_squared_error, r2_score


# 1. Load and clean the data
df = pd.read_csv("cars_hw.csv")

# Drop row index column
df = df.drop(columns=["Unnamed: 0"])

# Clean owner variable
owner_map = {"1st": 1, "2nd": 2, "3rd": 3}
df["Owners_Num"] = df["No_of_Owners"].map(owner_map)

# Engineer a few useful variables
df["Age"] = df["Make_Year"].max() + 1 - df["Make_Year"]
df["Log_Mileage"] = np.log(df["Mileage_Run"])

# Basic checks
print("Shape:", df.shape)
print("\nMissing values:\n", df.isna().sum())
print("\nData types:\n", df.dtypes)

# 2. EDA
print("\nSummary statistics:\n", df[["Price", "Mileage_Run", "Make_Year", "Seating_Capacity", "Owners_Num"]].describe())

print("\nAverage price by make:\n", df.groupby("Make")["Price"].mean().sort_values(ascending=False).head(10))
print("\nAverage price by fuel type:\n", df.groupby("Fuel_Type")["Price"].mean().sort_values(ascending=False))
print("\nAverage price by body type:\n", df.groupby("Body_Type")["Price"].mean().sort_values(ascending=False))

# Optional plots
fig, ax = plt.subplots(figsize=(6, 4))
ax.hist(df["Price"], bins=30)
ax.set_title("Distribution of Price")
ax.set_xlabel("Price")
ax.set_ylabel("Count")
plt.tight_layout()
plt.savefig("price_hist.png", dpi=150)
plt.close()

fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(df["Mileage_Run"], df["Price"], alpha=0.5)
ax.set_title("Price vs Mileage")
ax.set_xlabel("Mileage")
ax.set_ylabel("Price")
plt.tight_layout()
plt.savefig("price_vs_mileage.png", dpi=150)
plt.close()

# 3. Train/test split
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

X_train = train_df.drop(columns=["Price"])
X_test = test_df.drop(columns=["Price"])
y_train = train_df["Price"]
y_test = test_df["Price"]

def rmse(y_true, y_pred):
    return mean_squared_error(y_true, y_pred) ** 0.5

def evaluate_model(name, pipeline):
    pipeline.fit(X_train, y_train)
    pred_train = pipeline.predict(X_train)
    pred_test = pipeline.predict(X_test)
    out = {
        "model": name,
        "train_rmse": rmse(y_train, pred_train),
        "test_rmse": rmse(y_test, pred_test),
        "train_r2": r2_score(y_train, pred_train),
        "test_r2": r2_score(y_test, pred_test),
    }
    return out

# 4. Simple linear models
num_basic = ["Make_Year", "Mileage_Run", "Seating_Capacity", "Owners_Num"]
cat_small = ["Fuel_Type", "Transmission_Type", "Body_Type"]
cat_all = ["Make", "Color", "Body_Type", "Fuel_Type", "Transmission", "Transmission_Type"]

prep1 = ColumnTransformer([
    ("num", Pipeline([("imputer", SimpleImputer(strategy="median"))]), num_basic)
])

prep2 = ColumnTransformer([
    ("num", Pipeline([("imputer", SimpleImputer(strategy="median"))]), num_basic),
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("ohe", OneHotEncoder(drop="first", handle_unknown="ignore"))
    ]), cat_small)
])

prep3 = ColumnTransformer([
    ("num", Pipeline([("imputer", SimpleImputer(strategy="median"))]), num_basic),
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("ohe", OneHotEncoder(drop="first", handle_unknown="ignore"))
    ]), cat_all)
])

model1 = Pipeline([("prep", prep1), ("model", LinearRegression())])
model2 = Pipeline([("prep", prep2), ("model", LinearRegression())])
model3 = Pipeline([("prep", prep3), ("model", LinearRegression())])

results = []
results.append(evaluate_model("Numeric only", model1))
results.append(evaluate_model("Numeric + a few categorical", model2))
results.append(evaluate_model("All main effects", model3))

results_df = pd.DataFrame(results)
print("\nSimple model comparison:\n", results_df)

# 5. Partial correlation / CCPR plots
#    for the best simple model
simple_formula = """
Price ~ Make_Year + Mileage_Run + Seating_Capacity + Owners_Num
+ C(Make) + C(Color) + C(Body_Type) + C(Fuel_Type) + C(Transmission) + C(Transmission_Type)
"""

simple_sm = smf.ols(simple_formula, data=train_df).fit()

from statsmodels.graphics.regressionplots import plot_ccpr
for var in ["Make_Year", "Mileage_Run", "Seating_Capacity", "Owners_Num"]:
    fig = plt.figure(figsize=(6, 4))
    ax = fig.add_subplot(111)
    plot_ccpr(simple_sm, var, ax=ax)
    ax.set_title(f"CCPR Plot: {var}")
    plt.tight_layout()
    plt.savefig(f"{var}_ccpr.png", dpi=150)
    plt.close()

# 6. More complex model
complex_formula = """
np.log(Price) ~ Age + I(Age**2) + Log_Mileage + I(Log_Mileage**2)
+ Seating_Capacity + Owners_Num
+ C(Make) + C(Body_Type) + C(Fuel_Type) + C(Transmission_Type) + C(Transmission)
+ Age:C(Fuel_Type) + Log_Mileage:C(Transmission_Type) + C(Body_Type):C(Fuel_Type)
"""

complex_sm = smf.ols(complex_formula, data=train_df).fit()

complex_pred_train = np.exp(complex_sm.predict(train_df))
complex_pred_test = np.exp(complex_sm.predict(test_df))

complex_results = {
    "model": "Complex model",
    "train_rmse": rmse(y_train, complex_pred_train),
    "test_rmse": rmse(y_test, complex_pred_test),
    "train_r2": r2_score(y_train, complex_pred_train),
    "test_r2": r2_score(y_test, complex_pred_test),
}
print("\nComplex model results:\n", complex_results)

# 8. Lasso
eng_num = ["Age", "Log_Mileage", "Seating_Capacity", "Owners_Num"]
eng_cat = ["Make", "Body_Type", "Fuel_Type", "Transmission", "Transmission_Type", "Color"]

prep_lasso = ColumnTransformer([
    ("num", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("poly", PolynomialFeatures(degree=2, include_bias=False))
    ]), eng_num),
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("ohe", OneHotEncoder(drop="first", handle_unknown="ignore"))
    ]), eng_cat)
])

lasso_model = Pipeline([
    ("prep", prep_lasso),
    ("scale", StandardScaler(with_mean=False)),
    ("model", TransformedTargetRegressor(
        regressor=LassoCV(cv=5, random_state=42, max_iter=20000),
        func=np.log,
        inverse_func=np.exp
    ))
])

lasso_model.fit(X_train, y_train)
lasso_pred_train = lasso_model.predict(X_train)
lasso_pred_test = lasso_model.predict(X_test)

lasso_results = {
    "model": "Lasso",
    "train_rmse": rmse(y_train, lasso_pred_train),
    "test_rmse": rmse(y_test, lasso_pred_test),
    "train_r2": r2_score(y_train, lasso_pred_train),
    "test_r2": r2_score(y_test, lasso_pred_test),
}
print("\nLasso results:\n", lasso_results)

feature_names = lasso_model.named_steps["prep"].get_feature_names_out()
coefs = lasso_model.named_steps["model"].regressor_.coef_
selected = pd.Series(coefs, index=feature_names)
selected = selected[selected.abs() > 1e-8].sort_values(key=np.abs, ascending=False)
print("\nTop selected Lasso features:\n", selected.head(25))
```
