```python
1. A model is linear if it is linear in its parameters which are its coefficients.
2. The coefficient on a dummy variable represents the difference in the expected outcome between that category and the omitted baseline category, 
holding other variables constant; the intercept represents the expected outcome for the baseline group.
3. Linear regression can be used for classification, but it is not appropriate because it does not model probabilities correctly and can produce predictions
outside the [0,1] range.
4. Overfitting means a model is too tailored to the training data and doesn't generalize well so there is low training error but there would be higher test
or cross validation error.
5. Multicolinearity occurs when explanatory variables are highly correlated which makes it difficult to estimate their individual effects. Two-stage least
squares technique addresses this issue by using an instrumental variable that is correlated with the regressor but uncorrelated with the error term.
In the first stage, the problematic regressor is projected onto the instrument to obtain a predicted version that is free of correlation with the error.
In the second stage, the predicted variable replaces the original regressor in the regression which causes consistent estimates.
6. Nonlinear relationships can be modeled while keeping the model linear in parameters by transforming the inputs.
7. The intercept represents the expected value of y for the baseline omitted category, with all variables set to 0.
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

# Load and clean the data
df = pd.read_csv("./data/Q1_clean.csv").dropna().copy()

# Keep only positive prices so log(price) is defined
df = df[df["Price"] > 0].copy()

# Create log price
df["log_price"] = np.log(df["Price"])

# Average price and review score by borough
borough_summary = (
    df.groupby("Neighbourhood ")[["Price", "Review Scores Rating"]]
      .mean()
      .sort_values("Price", ascending=False)
)

print("\nAverage price and review score by borough:")
print(borough_summary)

# KDE plots of price and log(price) by borough
plt.figure(figsize=(9, 6))
for name, group in df.groupby("Neighbourhood "):
    group["Price"].plot(kind="kde", label=name)
plt.xlabel("Price")
plt.ylabel("Density")
plt.title("KDE of Price by Neighbourhood")
plt.legend()
plt.tight_layout()
plt.show()

plt.figure(figsize=(9, 6))
for name, group in df.groupby("Neighbourhood "):
    group["log_price"].plot(kind="kde", label=name)
plt.xlabel("log(Price)")
plt.ylabel("Density")
plt.title("KDE of log(Price) by Neighbourhood")
plt.legend()
plt.tight_layout()
plt.show()


# Price on Neighbourhood with no intercept
neigh_dummies_no_intercept = pd.get_dummies(
    df["Neighbourhood "],
    drop_first=False,
    dtype=float
)

model_no_intercept = sm.OLS(df["Price"], neigh_dummies_no_intercept).fit()

print("\nRegression of Price on Neighbourhood (no intercept):")
print(model_no_intercept.summary())
print("\nCoefficients (no intercept):")
print(model_no_intercept.params)

# Price on Neighbourhood with intercept
neigh_dummies_with_intercept = pd.get_dummies(
    df["Neighbourhood "],
    drop_first=True,
    dtype=float
)

X_with_intercept = sm.add_constant(neigh_dummies_with_intercept)
model_with_intercept = sm.OLS(df["Price"], X_with_intercept).fit()

print("\nRegression of Price on Neighbourhood (with intercept):")
print(model_with_intercept.summary())
print("\nCoefficients (with intercept):")
print(model_with_intercept.params)

# Train/test split for parts 4, 5, and 7
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

# Price: Review Scores Rating + Neighbourhood
X_train_4 = train_df[["Review Scores Rating"]].copy()
X_test_4 = test_df[["Review Scores Rating"]].copy()

train_neigh_4 = pd.get_dummies(train_df["Neighbourhood "], drop_first=True, dtype=float)
test_neigh_4 = pd.get_dummies(test_df["Neighbourhood "], drop_first=True, dtype=float)

X_train_4 = pd.concat([X_train_4, train_neigh_4], axis=1)
X_test_4 = pd.concat([X_test_4, test_neigh_4], axis=1)

# Make sure train/test columns match exactly
X_test_4 = X_test_4.reindex(columns=X_train_4.columns, fill_value=0)

X_train_4 = sm.add_constant(X_train_4)
X_test_4 = sm.add_constant(X_test_4, has_constant="add")

model_4 = sm.OLS(train_df["Price"], X_train_4).fit()
pred_4 = model_4.predict(X_test_4)

rmse_4 = mean_squared_error(test_df["Price"], pred_4, squared=False)
r2_4 = r2_score(test_df["Price"], pred_4)

print("\nPart 4: Price ~ Review Scores Rating + Neighbourhood")
print(model_4.summary())
print(f"Test RMSE: {rmse_4}")
print(f"Test R^2: {r2_4}")
print(f"Coefficient on Review Scores Rating: {model_4.params['Review Scores Rating']}")

# Price: Review Scores Rating + Neighbourhood + Property Type
X_train_5 = train_df[["Review Scores Rating"]].copy()
X_test_5 = test_df[["Review Scores Rating"]].copy()

train_neigh_5 = pd.get_dummies(train_df["Neighbourhood "], drop_first=True, dtype=float)
test_neigh_5 = pd.get_dummies(test_df["Neighbourhood "], drop_first=True, dtype=float)

train_prop_5 = pd.get_dummies(train_df["Property Type"], drop_first=True, dtype=float)
test_prop_5 = pd.get_dummies(test_df["Property Type"], drop_first=True, dtype=float)

X_train_5 = pd.concat([X_train_5, train_neigh_5, train_prop_5], axis=1)
X_test_5 = pd.concat([X_test_5, test_neigh_5, test_prop_5], axis=1)

# Match columns
X_test_5 = X_test_5.reindex(columns=X_train_5.columns, fill_value=0)

X_train_5 = sm.add_constant(X_train_5)
X_test_5 = sm.add_constant(X_test_5, has_constant="add")

model_5 = sm.OLS(train_df["Price"], X_train_5).fit()
pred_5 = model_5.predict(X_test_5)

rmse_5 = mean_squared_error(test_df["Price"], pred_5, squared=False)
r2_5 = r2_score(test_df["Price"], pred_5)

print("\nPart 5: Price ~ Review Scores Rating + Neighbourhood + Property Type")
print(model_5.summary())
print(f"Test RMSE: {rmse_5}")
print(f"Test R^2: {r2_5}")
print(f"Coefficient on Review Scores Rating: {model_5.params['Review Scores Rating']}")

# Find the most expensive property type coefficient in part 5
property_type_coefs = model_5.params[
    [col for col in model_5.params.index if col in train_prop_5.columns]
]

if len(property_type_coefs) > 0:
    print("\nProperty type coefficients from part 5:")
    print(property_type_coefs.sort_values(ascending=False))
    print(f"Most expensive property type relative to baseline: {property_type_coefs.idxmax()}")
    
# Price: Review Scores Rating + Neighbourhood * Property Type
train_interactions = pd.get_dummies(
    train_df[["Neighbourhood ", "Property Type"]],
    drop_first=True,
    dtype=float
)
test_interactions = pd.get_dummies(
    test_df[["Neighbourhood ", "Property Type"]],
    drop_first=True,
    dtype=float
)

# Build interaction terms manually
for neigh_col in [c for c in train_interactions.columns if c.startswith("Neighbourhood ")]:
    for prop_col in [c for c in train_interactions.columns if c.startswith("Property Type_")]:
        inter_col = f"{neigh_col}__X__{prop_col}"
        train_interactions[inter_col] = train_interactions[neigh_col] * train_interactions[prop_col]

for neigh_col in [c for c in test_interactions.columns if c.startswith("Neighbourhood ")]:
    for prop_col in [c for c in test_interactions.columns if c.startswith("Property Type_")]:
        inter_col = f"{neigh_col}__X__{prop_col}"
        test_interactions[inter_col] = test_interactions[neigh_col] * test_interactions[prop_col]

X_train_7 = pd.concat(
    [train_df[["Review Scores Rating"]].reset_index(drop=True), train_interactions.reset_index(drop=True)],
    axis=1
)
X_test_7 = pd.concat(
    [test_df[["Review Scores Rating"]].reset_index(drop=True), test_interactions.reset_index(drop=True)],
    axis=1
)

# Match columns
X_test_7 = X_test_7.reindex(columns=X_train_7.columns, fill_value=0)

X_train_7 = sm.add_constant(X_train_7)
X_test_7 = sm.add_constant(X_test_7, has_constant="add")

model_7 = sm.OLS(train_df["Price"].reset_index(drop=True), X_train_7).fit()
pred_7 = model_7.predict(X_test_7)

rmse_7 = mean_squared_error(test_df["Price"], pred_7, squared=False)
r2_7 = r2_score(test_df["Price"], pred_7)

print("\nPart 7: Price ~ Review Scores Rating + Neighbourhood * Property Type")
print(model_7.summary())
print(f"Test RMSE: {rmse_7}")
print(f"Test R^2: {r2_7}")
print(f"Coefficient on Review Scores Rating: {model_7.params['Review Scores Rating']}")

# Find the most expensive predicted combination in part 7
boroughs = sorted(df["Neighbourhood "].unique())
property_types = sorted(df["Property Type"].unique())
max_review = df["Review Scores Rating"].max()

candidate_rows = []
for borough in boroughs:
    for prop_type in property_types:
        row = {
            "Review Scores Rating": max_review,
            "Neighbourhood ": borough,
            "Property Type": prop_type
        }
        candidate_rows.append(row)

candidates = pd.DataFrame(candidate_rows)

candidate_dummies = pd.get_dummies(
    candidates[["Neighbourhood ", "Property Type"]],
    drop_first=True,
    dtype=float
)

for neigh_col in [c for c in candidate_dummies.columns if c.startswith("Neighbourhood ")]:
    for prop_col in [c for c in candidate_dummies.columns if c.startswith("Property Type_")]:
        inter_col = f"{neigh_col}__X__{prop_col}"
        candidate_dummies[inter_col] = candidate_dummies[neigh_col] * candidate_dummies[prop_col]

X_candidates = pd.concat(
    [candidates[["Review Scores Rating"]].reset_index(drop=True), candidate_dummies.reset_index(drop=True)],
    axis=1
)

X_candidates = X_candidates.reindex(columns=X_train_7.columns.drop("const"), fill_value=0)
X_candidates = sm.add_constant(X_candidates, has_constant="add")

candidates["predicted_price"] = model_7.predict(X_candidates)

print("\nMost expensive predicted borough/property-type combination in part 7:")
print(candidates.sort_values("predicted_price", ascending=False).head(10))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import OneHotEncoder, PolynomialFeatures, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.impute import SimpleImputer

# 1. Load and clean the data
df = pd.read_csv("./data/cars_hw.csv").copy()

# Drop index-like column
df = df.drop(columns=["Unnamed: 0"])

# Convert owners to numeric
df["No_of_Owners"] = df["No_of_Owners"].map({
    "1st": 1,
    "2nd": 2,
    "3rd": 3
}).astype(float)

# Merge rare fuel category
df["Fuel_Type"] = df["Fuel_Type"].replace({
    "petrol+cng": "petrol"
})

# Reduce skew in mileage
df["Mileage_Run_asinh"] = np.arcsinh(df["Mileage_Run"])

# Extract transmission speeds as numeric where possible
df["Transmission_Speeds"] = df["Transmission"].str.extract(r"(\d+)").astype(float)

# 2. Price summary and KDEs
print("Overall Price summary:")
print(df["Price"].describe())

print("\nPrice summary by Make:")
make_summary = df.groupby("Make")["Price"].describe().sort_values("mean", ascending=False)
print(make_summary)

# KDE of overall price
plt.figure(figsize=(8, 5))
df["Price"].plot(kind="kde")
plt.xlabel("Price")
plt.ylabel("Density")
plt.title("KDE of Car Prices")
plt.tight_layout()
plt.show()

# KDE by Make
plt.figure(figsize=(10, 6))
for name, grp in df.groupby("Make"):
    grp["Price"].plot(kind="kde", label=name, alpha=0.7)
plt.xlabel("Price")
plt.ylabel("Density")
plt.title("KDE of Price by Make")
plt.legend(fontsize=7, ncol=2)
plt.tight_layout()
plt.show()

# 3. Train / test split
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

target = "Price"

numeric_features = [
    "Make_Year",
    "Seating_Capacity",
    "No_of_Owners",
    "Mileage_Run_asinh",
    "Transmission_Speeds"
]

categorical_features = [
    "Make",
    "Color",
    "Body_Type",
    "Fuel_Type",
    "Transmission",
    "Transmission_Type"
]

def rmse(y_true, y_pred):
    return mean_squared_error(y_true, y_pred) ** 0.5

def fit_eval(train_df, test_df, num_feats, cat_feats, poly_degree=None):
    transformers = []

    if num_feats:
        num_steps = [
            ("imputer", SimpleImputer(strategy="median"))
        ]

        if poly_degree is not None:
            num_steps += [
                ("scaler", StandardScaler()),
                ("poly", PolynomialFeatures(degree=poly_degree, include_bias=False))
            ]

        transformers.append((
            "num",
            Pipeline(num_steps),
            num_feats
        ))

    if cat_feats:
        cat_steps = [
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("onehot", OneHotEncoder(drop="first", handle_unknown="ignore"))
        ]
        transformers.append((
            "cat",
            Pipeline(cat_steps),
            cat_feats
        ))

    preprocessor = ColumnTransformer(transformers)

    model = Pipeline([
        ("pre", preprocessor),
        ("lr", LinearRegression())
    ])

    X_train = train_df[num_feats + cat_feats]
    y_train = train_df[target]

    X_test = test_df[num_feats + cat_feats]
    y_test = test_df[target]

    model.fit(X_train, y_train)

    pred_train = model.predict(X_train)
    pred_test = model.predict(X_test)

    results = {
        "r2_train": r2_score(y_train, pred_train),
        "r2_test": r2_score(y_test, pred_test),
        "rmse_train": rmse(y_train, pred_train),
        "rmse_test": rmse(y_test, pred_test),
        "pred_train": pred_train,
        "pred_test": pred_test
    }

    return model, results

# 4. Numeric-only model
model_num, res_num = fit_eval(
    train_df=train_df,
    test_df=test_df,
    num_feats=numeric_features,
    cat_feats=[]
)

print("\nNumeric-only model")
print(f"Train R^2:  {res_num['r2_train']:.4f}")
print(f"Test R^2:   {res_num['r2_test']:.4f}")
print(f"Train RMSE: {res_num['rmse_train']:.2f}")
print(f"Test RMSE:  {res_num['rmse_test']:.2f}")

# 5. Categorical-only model
model_cat, res_cat = fit_eval(
    train_df=train_df,
    test_df=test_df,
    num_feats=[],
    cat_feats=categorical_features
)

print("\nCategorical-only model")
print(f"Train R^2:  {res_cat['r2_train']:.4f}")
print(f"Test R^2:   {res_cat['r2_test']:.4f}")
print(f"Train RMSE: {res_cat['rmse_train']:.2f}")
print(f"Test RMSE:  {res_cat['rmse_test']:.2f}")

# 6. Combined model
model_comb, res_comb = fit_eval(
    train_df=train_df,
    test_df=test_df,
    num_feats=numeric_features,
    cat_feats=categorical_features
)

print("\nCombined model")
print(f"Train R^2:  {res_comb['r2_train']:.4f}")
print(f"Test R^2:   {res_comb['r2_test']:.4f}")
print(f"Train RMSE: {res_comb['rmse_train']:.2f}")
print(f"Test RMSE:  {res_comb['rmse_test']:.2f}")

print("\nImprovement of combined over categorical-only:")
print(f"Delta test R^2:  {res_comb['r2_test'] - res_cat['r2_test']:.4f}")
print(f"Delta test RMSE: {res_cat['rmse_test'] - res_comb['rmse_test']:.2f}")

# 7. Polynomial expansion on numeric features
poly_results = []

for degree in range(1, 7):
    _, res_poly = fit_eval(
        train_df=train_df,
        test_df=test_df,
        num_feats=numeric_features,
        cat_feats=[],
        poly_degree=degree
    )

    poly_results.append({
        "degree": degree,
        "train_r2": res_poly["r2_train"],
        "test_r2": res_poly["r2_test"],
        "train_rmse": res_poly["rmse_train"],
        "test_rmse": res_poly["rmse_test"]
    })

poly_df = pd.DataFrame(poly_results)
print("\nPolynomial results:")
print(poly_df)

# 8. Best model diagnostics
# Best overall model = combined model
y_test = test_df["Price"].values
pred_test = res_comb["pred_test"]
residuals = y_test - pred_test

# Predicted vs true
plt.figure(figsize=(6, 6))
plt.scatter(y_test, pred_test, alpha=0.6)
mn = min(y_test.min(), pred_test.min())
mx = max(y_test.max(), pred_test.max())
plt.plot([mn, mx], [mn, mx], linestyle="--")
plt.xlabel("True Price")
plt.ylabel("Predicted Price")
plt.title("Predicted vs True Price (Best Model)")
plt.tight_layout()
plt.show()

# Residual KDE
plt.figure(figsize=(8, 5))
pd.Series(residuals).plot(kind="kde")
plt.axvline(0, linestyle="--")
plt.xlabel("Residual (True - Predicted)")
plt.ylabel("Density")
plt.title("Residual KDE on Test Set (Best Model)")
plt.tight_layout()
plt.show()
```

**Q4.** Let's explore multiple linear regression in a two-variable case, to build more intuition about what is happening.

Suppose the model is 
$$
\hat{y}_i = b_0 + b_1 z_{i1} + b_2 z_{i2}
$$
Assume that $z_{ij}$ is centered or de-meaned, so that $z_{ij} = x_{ij} - m_j$ where $m_j$ is the mean of variable $j$ and $x_{ij}$ is the original value of variable $j$ for observation $i$. Notice that this implies
$$
\dfrac{1}{N} \sum_{i=1}^N z_{ij} = 0
$$
which will simplify your calculations below substantially!

1. Write down the SSE for this model.
2. Take partial derivatives with respect to $b_0$, $b_1$, and $b_2$.
3. Verify that the average error is zero and $e \cdot z =0$ at the optimum, just as in the single linear regression case.
4. Show that the optimal intercept is $b_0^* = \bar{y}$. Eliminate $b_0^*$ from the remaining equations, and focus on $b_1$ and $b_2$.
5. Write your results as a matrix equation in the form "$Ab=C$". These are called the **normal equations**.
6. Divide both sides by $N$ and substitute $z_{ij} = x_{ij} - m_j$ back into your normal equations for $x_{ij}$. What is the matrix $A$? What is the vector $C$? Explain the intuition of your discovery.

1. The SSE is
$$
SSE(b_0, b_1, b_2) =
\sum_{i=1}^N \left(y_i - b_0 - b_1 z_{i1} - b_2 z_{i2}\right)^2
$$

2. Partial Derivatives

Let $e_i = y_i - b_0 - b_1 z_{i1} - b_2 z_{i2}$.

With respect to $b_0$:
$$
\frac{\partial SSE}{\partial b_0} = -2 \sum_{i=1}^N e_i
$$

With respect to $b_1$:
$$
\frac{\partial SSE}{\partial b_1} = -2 \sum_{i=1}^N z_{i1} e_i
$$

With respect to $b_2$:
$$
\frac{\partial SSE}{\partial b_2} = -2 \sum_{i=1}^N z_{i2} e_i
$$

3. First-Order Conditions
At the optimum:

$$
\sum_{i=1}^N e_i = 0
$$

$$
\sum_{i=1}^N z_{i1} e_i = 0, \quad \sum_{i=1}^N z_{i2} e_i = 0
$$

Interpretation:
The average residual is zero.
Residuals are orthogonal to each regressor.


4. Solve for the Intercept

From:

$$
\sum_{i=1}^N (y_i - b_0 - b_1 z_{i1} - b_2 z_{i2}) = 0
$$

Using $\sum z_{ij} = 0$:

$$
\sum y_i - N b_0 = 0
$$

$$
b_0^* = \bar{y}
$$


5. Normal Equations (Matrix Form)

After substituting $b_0 = \bar{y}$:

$$
\begin{bmatrix}
\sum z_{i1}^2 & \sum z_{i1} z_{i2} \\
\sum z_{i1} z_{i2} & \sum z_{i2}^2
\end{bmatrix}
\begin{bmatrix}
b_1 \\
b_2
\end{bmatrix}
=
\begin{bmatrix}
\sum z_{i1} (y_i - \bar{y}) \\
\sum z_{i2} (y_i - \bar{y})
\end{bmatrix}
$$

6. Divide by $N$ and Express in Terms of $x$

Since $z_{ij} = x_{ij} - m_j$:

$$
\frac{1}{N} \sum z_{i1}^2 = \mathrm{Var}(x_1)
$$

$$
\frac{1}{N} \sum z_{i2}^2 = \mathrm{Var}(x_2)
$$

$$
\frac{1}{N} \sum z_{i1} z_{i2} = \mathrm{Cov}(x_1, x_2)
$$

$$
\frac{1}{N} \sum z_{i1}(y_i - \bar{y}) = \mathrm{Cov}(x_1, y)
$$

$$
\frac{1}{N} \sum z_{i2}(y_i - \bar{y}) = \mathrm{Cov}(x_2, y)
$$

Final Result

$$
\begin{bmatrix}
\mathrm{Var}(x_1) & \mathrm{Cov}(x_1, x_2) \\
\mathrm{Cov}(x_1, x_2) & \mathrm{Var}(x_2)
\end{bmatrix}
\begin{bmatrix}
b_1 \\
b_2
\end{bmatrix}
=
\begin{bmatrix}
\mathrm{Cov}(x_1, y) \\
\mathrm{Cov}(x_2, y)
\end{bmatrix}
$$


Intuition

- The matrix on the left captures relationships between predictors.
- The vector on the right captures relationships with the outcome.

This shows that regression balances:
- how each variable relates to $y$
- how the variables relate to each other

High correlation between predictors leads to instability in coefficients (multicollinearity).
